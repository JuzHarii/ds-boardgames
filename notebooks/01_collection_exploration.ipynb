  - Người dùng tích cực trên BGG chủ yếu là người chơi lâu năm và có gu chơi game khá sâu (hobbyist gamers). Nhóm người dùng này thường có xu hướng ưu ái các tựa game chiến thuật phức tạp, nên điểm số chưa chắc đã đại diện hoàn toàn cho sở thích của người chơi giải trí đại chúng.
