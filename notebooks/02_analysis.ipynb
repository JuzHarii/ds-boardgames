Mỗi câu hỏi phi-ML phải áp dụng ít nhất **2 trong 6 tiêu chí độ sâu**, và toàn bộ các câu hỏi gộp lại phải bao phủ ít nhất **4/6 tiêu chí**:
- **D1 (Combining data)**: Gộp thêm bảng dữ liệu/nguồn ngoài (kiểm tra row count trước/sau, kiểm tra unmatched keys).
- **D2 (Reshaping & time)**: Sử dụng pivot, melt, resampling, rolling windows, lags, so sánh qua từng thời kỳ/chu kỳ.
